```python
import pandas as pd

# Load the dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

# Show dataset size
print("Number of records:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display first 5 rows
df.head()
```

```python
# Check all column names and data types

print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Handle missing values using median

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# Check again
print("Total missing values after cleaning:", df.isnull().sum().sum())
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
# Check unique values in categorical columns

categorical_columns = [
    'State',
    'District',
    'Crop',
    'Season',
    'Irrigation_Method'
]

for column in categorical_columns:
    print(f"\n{column}:")
    print("Number of unique values:", df[column].nunique())
    print(df[column].unique())
```

```python
Q1 = sugarcane_yield.quantile(0.25)
Q3 = sugarcane_yield.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = sugarcane_yield[
    (sugarcane_yield < lower_limit) |
    (sugarcane_yield > upper_limit)
]

print("\nNumber of Sugarcane Yield Outliers:", len(outliers))
print("\nOutlier Values:")
print(outliers)
```

```python
# Step 98.2 - Select features and target

features = [
    'Crop',
    'Season',
    'Farm_Area_Hectares',
    'Production_Tonnes',
    'Yield_Tonnes_Ha',
    'Revenue_INR',
    'Total_Cost_INR'
]

target = 'Profit_INR'

X = df[features].copy()
y = df[target].copy()

print("Features:")
print(features)

print("\nTarget:")
print(target)

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Step 98.3 - Check selected data

print("Missing values in features:")
print(X.isnull().sum())

print("\nMissing values in target:")
print(y.isnull().sum())

print("\nFeature data types:")
print(X.dtypes)
```

```python
# Step 99.1 - One-hot encode categorical columns

X_encoded = pd.get_dummies(
    X,
    columns=['Crop', 'Season'],
    drop_first=True
)

print("Encoded feature shape:", X_encoded.shape)

print("\nEncoded columns:")
print(X_encoded.columns.tolist())

display(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("Training features:", X_train.shape[1])
print("Testing features:", X_test.shape[1])
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)

linear_predictions = linear_model.predict(X_test)

print("Linear Regression training completed.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("===== LINEAR REGRESSION =====")
print("MAE :", linear_mae)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train,
    y_train
)

rf_predictions = rf_model.predict(X_test)

print("Random Forest training completed.")
```

```python
rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("===== RANDOM FOREST =====")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_model.fit(
    X_train,
    y_train
)

gb_predictions = gb_model.predict(X_test)

print("Gradient Boosting training completed.")
```

```python
gb_mae = mean_absolute_error(
    y_test,
    gb_predictions
)

gb_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        gb_predictions
    )
)

gb_r2 = r2_score(
    y_test,
    gb_predictions
)

print("===== GRADIENT BOOSTING =====")
print("MAE :", gb_mae)
print("RMSE:", gb_rmse)
print("R²  :", gb_r2)
```

```python
model_comparison = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Random Forest',
        'Gradient Boosting'
    ],
    'MAE': [
        linear_mae,
        rf_mae,
        gb_mae
    ],
    'RMSE': [
        linear_rmse,
        rf_rmse,
        gb_rmse
    ],
    'R2_Score': [
        linear_r2,
        rf_r2,
        gb_r2
    ]
})

model_comparison = model_comparison.sort_values(
    'R2_Score',
    ascending=False
).reset_index(drop=True)

display(model_comparison)
```

```python
# Step 115 - Reusable Profit Prediction Function

def predict_profit(
    crop,
    season,
    farm_area,
    production,
    yield_per_ha,
    revenue,
    total_cost
):
    # Create input dataframe
    input_data = pd.DataFrame({
        'Farm_Area_Hectares': [farm_area],
        'Production_Tonnes': [production],
        'Yield_Tonnes_Ha': [yield_per_ha],
        'Revenue_INR': [revenue],
        'Total_Cost_INR': [total_cost]
    })

    # Add categorical columns
    for column in X_encoded.columns:
        if column.startswith('Crop_'):
            input_data[column] = 0

        if column.startswith('Season_'):
            input_data[column] = 0

    # Set selected crop
    crop_column = 'Crop_' + crop

    if crop_column in input_data.columns:
        input_data[crop_column] = 1

    # Set selected season
    season_column = 'Season_' + season

    if season_column in input_data.columns:
        input_data[season_column] = 1

    # Match training columns exactly
    input_data = input_data.reindex(
        columns=X_encoded.columns,
        fill_value=0
    )

    # Predict
    prediction = best_model.predict(input_data)[0]

    return prediction


# Test prediction
prediction = predict_profit(
    crop='Chilli',
    season='Rabi',
    farm_area=10,
    production=25,
    yield_per_ha=2.5,
    revenue=500000,
    total_cost=300000
)

print("======================================")
print("       PROFIT PREDICTION")
print("======================================")
print(f"Predicted Profit: ₹{prediction:,.2f}")
```
